```python
import numpy as np

# A toy "counts matrix": 3 cells (rows) x 4 genes (columns)
counts = np.array([
    [0, 3, 1, 0],
    [2, 0, 0, 5],
    [1, 1, 4, 0],
])

print(counts)
print("shape:", counts.shape)   # (3, 4)  -> (n_cells, n_genes)
print("ndim: ", counts.ndim)    # 2       -> two dimensions (it's a matrix)
print("size: ", counts.size)    # 12      -> total number of values
print("dtype:", counts.dtype)   # int64   -> every value shares ONE type
```
